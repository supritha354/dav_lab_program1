```python
import pandas as pd
import numpy as np

# Sample DataFrame to define 'df' for demonstration purposes
data = {'Final_Score': np.random.randint(50, 100, 100)}
df = pd.DataFrame(data)

# Add the missing columns with sample data
df['Attendence'] = np.random.randint(70, 100, 100) # Assuming attendance percentage
df['Assignment_Score'] = np.random.randint(60, 100, 100)
df['Internal_Marks'] = np.random.randint(50, 95, 100)
df['Study_Hours'] = np.random.randint(1, 15, 100) # Adding Study_Hours column

plt.figure(figsize=(8,5))
plt.hist(df["Final_Score"],bins=6,edgecolor="green")
plt.title("Distribution of Final Scores")
plt.xlabel("Final Score"); plt.ylabel("Number of Students")
plt.show()
```

```python
import numpy as np

# Add a sample 'Department' column to df for demonstration
departments = ['Math', 'Science', 'English', 'History']
df['Department'] = np.random.choice(departments, size=len(df))

plt.figure(figsize=(8,5))
sns.boxplot(data=df,x="Department",y="Final_Score")
plt.title("Final Score distribution by Department")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x="Study_Hours",y="Final_Score",hue="Department",s=100)
plt.title("Study hours vs Final Score by Department")
plt.show()
```

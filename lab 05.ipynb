```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import zscore
data ={
    "Students":["A","B","C","D","E","F","G","H",
                "I","J","K","L","M","N","O"],
    "Study_Hours":[2,3,4,5,4,6,5,3,4,5,6,6,4,5,30]

}
df = pd.DataFrame(data)
print("Original Dataset")
display(df)
```

```python
df["Study_Hours"].describe()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=df["Study_Hours"])
plt.title("Study Hours - Box Plot")
plt.xlabel("Study Hours")
plt.show()
```

```python
Q1 = df["Study_Hours"].quantile(0.25)
Q3 = df["Study_Hours"].quantile(0.75)
print("Q1:",Q1)
print("Q3:",Q3)
```

```python
IQR =Q3 - Q1
print ("IQR:",IQR)
```

```python
lower_bound = Q1 - 1.5*IQR
upper_bound = Q3 + 1.5*IQR
print("Lower Bound:",lower_bound)
print("Upper Bound:",upper_bound)
```

```python
normal_data_iqr = df[
    (df["Study_Hours"] >= lower_bound) &
    (df["Study_Hours"] <= upper_bound)
]
display(normal_data_iqr)
```

```python
df_capped = df.copy()
df_capped["Study_Hours"] = df_capped["Study_Hours"].clip(
    lower = lower_bound,
    upper = upper_bound
)
display(df_capped)
```

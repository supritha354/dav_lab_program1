```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
```

```python
df = pd.read_csv('housing_clean.csv')
```

```python
X_multi = df[['Area','Rooms','Age','Location_Score']]
x_tr2, x_te2, y_tr2, y_te2 = train_test_split(
    X_multi, df['Price'], test_size=0.2, random_state=42
)
model_multi = LinearRegression()
model_multi.fit(x_tr2, y_tr2)
y_pred_multi = model_multi.predict(x_te2)
print('MULTIPLE REGRESSION')
print('R^2', round(r2_score(y_te2, y_pred_multi), 2))
print('RMSE', round(np.sqrt(mean_squared_error(y_te2, y_pred_multi)), 2))
```

```python
residuals = y_te2 - y_pred_multi
plt.scatter(y_pred_multi, residuals, alpha=0.5)
plt.axhline(y=0, color='red', linestyle='--')
plt.xlabel('Predicted Price')
plt.ylabel('Residuals')
plt.title('Residual Plot')
plt.show()
```
